```python
import pandas as pd
import numpy as np

columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 
           'acceleration', 'model_year', 'origin', 'car_name']

df = pd.read_csv(r'../DATA/RAWDATA/auto-mpg.data' , sep=r'\s+', names=columns, na_values='?')
df.head()
```

```python
df['weight'].mean()
```

```python

print("\nDESCRIBE:")
print(df.describe())
```

```python
#checking missing values
df.isnull().sum() 

```

```python
df['horsepower'].isnull().sum() 
df[df['horsepower'].isnull()]

```

```python
print("--- MINIMUM VALUES ---")
print(df.min(numeric_only=True))

print("\n--- MAXIMUM VALUES ---")
print(df.max(numeric_only=True))
```

```python
df['model_year']=df['model_year'] + 1900
df.head()
```

```python

df['model_year'] = df['model_year'] + 3620



np.random.seed(42)
df['origin'] = np.random.choice(['USA', 'Europe', 'Japan','Asia','Africa'], size=len(df))

df.head()
```

```python
# NumPy array calculation
mpg_array = df["mpg"].to_numpy()
```

```python
df.head()
```

```python

# mean median max min for mpg
print('--- OVERALL DATASET SUMMARY ---')

print(df[['mpg', 'horsepower', 'weight', 'acceleration']].agg(['mean', 'median', 'min', 'max','std']).round(2))

```

```python
#mean of largest mpg
df.groupby('model_year')['mpg'].mean().nlargest(1)
```

```python
#compare avg mpg by origin
df.groupby('origin')['mpg'].mean()
```

```python

df['avg_mpg_by_cyl'] = df.groupby('cylinders')['mpg'].transform('mean')
df = df[df['cylinders'].isin([4, 6, 8])]
df.head()
```

```python
#compare avg mpg for cars with 4,6,8 cylinders
df[df['cylinders'].isin([4,6,8])].groupby('cylinders')['mpg'].mean()
```

```python
df.duplicated().sum()
```

```python

high_mpg = df[df['mpg'] > 30]
print(high_mpg.head())
print(high_mpg.info())
df.head()


```

```python

df['power_to_weight'] = df['horsepower'] / df['weight']


df['power_to_weight'] = df['power_to_weight'].round(4)
```

```python
# Top 10 fuel-efficient cars

print("\nTOP 10 FUEL-EFFICIENT CARS:")
print(df.sort_values("mpg", ascending=False)
      [["mpg", "cylinders", "horsepower", "model_year", "car_name"]]
      .head(10))
```

```python
# Cylinder-count bar chart
df["cylinders"].value_counts().sort_index().plot(
    kind="bar", title="Number of Cars by Cylinders"
)
plt.xlabel("Cylinders")
plt.ylabel("Number of Cars")
plt.tight_layout()
plt.show()
```

```python
# MPG histogram
plt.hist(df["mpg"], bins=15)
plt.title("Distribution of MPG")
plt.xlabel("Miles per Gallon (MPG)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
